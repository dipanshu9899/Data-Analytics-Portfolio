```python
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    database="End-to-End",
    user="postgres",
    password="1234",
    port="5432"
)

print("Database connected successfully!")
```

Customer

```python
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE customers (
    customer_id VARCHAR(10) PRIMARY KEY,
    customer_name VARCHAR(100),
    gender VARCHAR(20),
    age INT,
    city VARCHAR(50),
    state VARCHAR(50),
    signup_date DATE
);
""")

conn.commit()

print("Customers table created successfully!")
```

```python
from psycopg2.extras import execute_values

data = list(customers.itertuples(index=False, name=None))

insert_query = """
INSERT INTO customers
(customer_id, customer_name, gender, age, city, state, signup_date)
VALUES %s
"""

execute_values(cursor, insert_query, data)
conn.commit()

print("Customers data imported successfully!")
```

```python
cursor.execute("SELECT COUNT(*) FROM customers;")

print(cursor.fetchone())
```

Product

```python
cursor.execute("""
CREATE TABLE products (
    product_id VARCHAR(10) PRIMARY KEY,
    product_name VARCHAR(100),
    category VARCHAR(50),
    sub_category VARCHAR(50),
    price NUMERIC(10,2),
    cost NUMERIC(10,2)
);
""")

conn.commit()

print("Products table created successfully!")
```

```python
data = list(products.itertuples(index=False, name=None))

insert_query = """
INSERT INTO products
(product_id, product_name, category, sub_category, price, cost)
VALUES %s
"""

execute_values(cursor, insert_query, data)
conn.commit()

print("Products data imported successfully!")
```

```python
cursor.execute("SELECT COUNT(*) FROM products;")

print(cursor.fetchone())
```

Orders

```python
cursor.execute("""
CREATE TABLE orders (
    order_id VARCHAR(10) PRIMARY KEY,
    customer_id VARCHAR(10),
    order_date DATE,
    product_id VARCHAR(10),
    quantity INT,
    discount NUMERIC(5,2),
    payment_method VARCHAR(30),
    region VARCHAR(20)
);
""")

conn.commit()

print("Orders table created successfully!")
```

```python
data = list(orders.itertuples(index=False, name=None))

insert_query = """
INSERT INTO orders
(order_id, customer_id, order_date, product_id, quantity,
 discount, payment_method, region)
VALUES %s
"""

execute_values(cursor, insert_query, data)
conn.commit()

print("Orders data imported successfully!")
```

```python
cursor.execute("SELECT COUNT(*) FROM orders;")

print(cursor.fetchone())
```

Return

```python
cursor.execute("""
CREATE TABLE returns (
    return_id VARCHAR(10) PRIMARY KEY,
    order_id VARCHAR(10),
    return_date DATE,
    return_reason VARCHAR(50)
);
""")

conn.commit()

print("Returns table created successfully!")
```

```python
data = list(returns.itertuples(index=False, name=None))

insert_query = """
INSERT INTO returns
(return_id, order_id, return_date, return_reason)
VALUES %s
"""

execute_values(cursor, insert_query, data)
conn.commit()

print("Returns data imported successfully!")
```

```python
cursor.execute("SELECT COUNT(*) FROM returns;")

print(cursor.fetchone())
```
