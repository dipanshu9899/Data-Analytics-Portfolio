PhonePe par kaunsi service sabse zyada use ho rahi hai aur kaunsi service sabse zyada transaction value generate kar rahi hai?

Service-wise transaction count

```python
service_tranction=transactions['Service'].value_counts()
service_tranction
```

Service-wise total amount:

```python
service_amount=transactions.groupby("Service")['Amount'].sum().sort_values(ascending=False)
service_amount
```

Average Transaction Value by Service

```python
avg_transaction_value=transactions.groupby('Service')['Amount'].mean().sort_values(ascending=False)
avg_transaction_value
```

Combine for Both Table

```python
service_summary = transactions.groupby("Service").agg(
    Transaction_Count=("Transaction_ID", "count"),
    Total_Amount=("Amount", "sum"),
    Average_Amount=("Amount", "mean")
).sort_values("Total_Amount", ascending=False)

service_summary
```

Service-wise Failure Rate

```python
service_failure = transactions.groupby("Service").agg(
    Total_Transactions=("Transaction_ID", "count"),
    Failed_Transactions=("Transaction_Result", lambda x: (x == "Failed").sum())
)

service_failure["Failure_Rate_%"] = (
    service_failure["Failed_Transactions"]
    / service_failure["Total_Transactions"]
) * 100

service_failure.sort_values("Failure_Rate_%", ascending=False)
```

Ab Customer Analysis

Which customer active in our platform

```python
user_transaction_count = transactions.groupby("User_ID").agg(
    Transaction_Count=("Transaction_ID", "count"),
    Total_Amount=("Amount", "sum"),
    Average_Amount=("Amount", "mean")
).sort_values("Transaction_Count", ascending=False)

user_transaction_count.head(10)
```

Top Customers by Transaction Value

```python
Top_Value_customer = transactions.groupby("User_ID").agg(
    Transaction_Count=("Transaction_ID", "count"),
    Total_Amount=("Amount", "sum"),
    Average_Amount=("Amount", "mean")
).sort_values("Total_Amount", ascending=False)
Top_Value_customer.head(10)
```

Customer + User Data Combine

```python
customer_analysis = users.merge(
    user_transaction_count,
    on="User_ID",
    how="left"
)
```

```python
customer_analysis.head()
```

Create Age Group 

```python
customer_analysis["Age_Group"] = customer_analysis["Age"].apply(
    lambda x:
        "18-25" if x <= 25
        else "26-35" if x <= 35
        else "36-45" if x <= 45
        else "46-55" if x <= 55
        else "56-60"
)
```

```python
customer_analysis["Age_Group"].value_counts()
```

Kaunsi age group actually sabse zyada transactions aur transaction value generate karti hai?

```python
age_group_analysis = customer_analysis.groupby("Age_Group").agg(
    Customers=("User_ID", "count"),
    Total_Transactions=("Transaction_Count", "sum"),
    Total_Amount=("Total_Amount", "sum"),
    Average_Transaction=("Average_Amount", "mean")
).sort_values("Total_Amount", ascending=False)

age_group_analysis
```

Inactive users ka join-date check

```python
inactive_users=customer_analysis[
    customer_analysis["Transaction_Count"].isna()
]
inactive_users['Join_Date'].describe()
```

User Tenure Analysis

```python
customer_analysis['Has_Transaction']=customer_analysis[
    "Transaction_Count"
].notna().map({
    True:"Yes",
    False:"No"
})
```

Join Year ke according transaction activity

```python
join_year_activity = customer_analysis.groupby(
    customer_analysis["Join_Date"].dt.year
).agg(
    Total_Users=("User_ID", "count"),
    Users_With_Transaction=("Has_Transaction", lambda x: (x == "Yes").sum()),
    Users_Without_Transaction=("Has_Transaction", lambda x: (x == "No").sum())
)

join_year_activity
```

Customer Segmentation

```python
active_customer=customer_analysis[
    customer_analysis['Has_Transaction']=='Yes'
].copy()
active_customer['Transaction_Count']=active_customer[
    "Transaction_Count"
].fillna(0)
active_customer['Total_Amount']=active_customer[
    "Total_Amount"
].fillna(0)
active_customer.shape
```

```python
active_customer['Customer_Value_Segment']=pd.qcut(
    active_customer['Total_Amount'],
    q=3,
    labels=['Low_Value','Medium_Value','High_Value']
)
active_customer['Customer_Value_Segment'].value_counts()
```

Segment ka actual business behavior

```python
segment_anaysis=active_customer.groupby(
    "Customer_Value_Segment"
).agg(
    customers=("User_ID","count"),
    Total_Transactions=("Transaction_Count","sum"),
    Total_Amount=('Total_Amount','sum'),
    Average_Transaction=('Average_Amount','mean')
)
segment_anaysis
```

Customer Frequency Segmentation

```python
active_customer['Frequency_segment']=pd.qcut(
    active_customer['Transaction_Count'],
    q=3,
    labels=['Low_Frequency','Medium_Frequency','High_Frequency']
)
active_customer['Frequency_segment'].value_counts()
```

Frequency Segment ka performance

```python
frequency_analysis = active_customer.groupby(
    "Frequency_segment",
    observed=True
).agg(
    Customers=("User_ID", "count"),
    Total_Transactions=("Transaction_Count", "sum"),
    Total_Amount=("Total_Amount", "sum"),
    Average_Transaction=("Average_Amount", "mean")
)

frequency_analysis
```

High Value customers aur High Frequency customers kitne overlap karte hain?

```python
customer_segments=active_customer.groupby(
    ['Customer_Value_Segment','Frequency_segment'],
    observed=True
).size().unstack(fill_value=0)
customer_segments
```

Our most valuable customers kis age group mein concentrated hain

```python
hv_hf_customers = active_customer[
    (active_customer["Customer_Value_Segment"] == "High_Value") &
    (active_customer["Frequency_segment"] == "High_Frequency")
]

hv_hf_age = hv_hf_customers["Age_Group"].value_counts().sort_index()

hv_hf_age
```
